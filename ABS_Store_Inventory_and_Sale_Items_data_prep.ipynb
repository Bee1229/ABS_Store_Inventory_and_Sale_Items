```python
import pandas as pd 
import numpy as np 
```

```python
data = "ABS_Store_Inventory_and_Sale_Items_20261104.csv"
```

```python
abs_store = pd.read_csv(data)

abs_store
```

# Missing Values

```python
abs_store.isnull().sum()
```

I don't consider this as misssing values because at that time the items were not sales.

```python
abs_store['On_Sales'] = abs_store['Sale Price'].notna()

abs_store
```

# Duplicate Value

```python
abs_store.duplicated().sum()
```

# Standadizing the Columns

```python
abs_store.columns
```

```python
abs_store.columns = abs_store.columns.str.replace(' ', '_').str.lower()


abs_store.columns
```

```python
columns = ['price','sale_price']

for i in columns:
    abs_store[i] = abs_store[i].str.replace(r"[$,]", "", regex=True)


abs_store
```

# Data Type Confirmation

```python
columns = ['price','sale_price']

for c in columns:
    abs_store[c]= pd.to_numeric(abs_store[c])

abs_store['sale_end_date'] = pd.to_datetime(abs_store['sale_end_date'])

abs_store.info()
```

# Outlier Detection

```python
abs_store['total_inventory'].quantile([0.25,0.75])
```

```python
abs_store['price'].quantile([0.25,0.75])
```

```python
abs_store['sale_price'].quantile([0.25,0.75])
```

### Additional column to group the total inventory

```python
#Total inventroy = 0 is Zero
#Total inventory < 100 is low
#Total inventory => 100 is Healthy
abs_store["stock_status"] = np.select(
    [abs_store["total_inventory"] == 0, abs_store["total_inventory"] < 100],
    ["Zero", "Low"],
    default="Healthy",
)

abs_store.head()
```

```python
abs_store['stock_status'].value_counts().unique
```
