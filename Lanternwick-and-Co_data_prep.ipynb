```python
import pandas as pd 
import numpy as np 
```

```python
data = "Lanternwick-and-Co.xlsx"
```

```python
lanternwick = pd.read_excel(data)

lanternwick.head()
```

```python
lanternwick.info()
```

```python
lanternwick.describe()
```

# Missing Values

```python
lanternwick.isnull().sum()
```

```python
columns = ['Description', 'CustomerID']

for i in columns:
    lanternwick[i] = lanternwick[i].fillna('None')

lanternwick.head()
```

```python
lanternwick.isnull().sum()
```

# Duplicate Value

```python
lanternwick.duplicated().sum()
```

```python
lanternwick = lanternwick.drop_duplicates()
```

```python
lanternwick.duplicated().sum()
```

# Standadizing the Columns

```python
lanternwick.columns
```

```python
lanternwick.columns = lanternwick.columns.str.replace(' ', '_').str.lower()


lanternwick.columns
```

# Data Type Confirmation

No data conversion needed.

# Outlier Detection

```python
lanternwick['quantity'].quantile([0.25,0.75])
```

```python
lanternwick['unitprice'].quantile([0.25,0.75])
```

### Additional column to group the total inventory

```python
lanternwick['linetotal'] = lanternwick['quantity'] * lanternwick['unitprice']

lanternwick.head()
```

# Saving the cleaned up table

```python
lanternwick.to_csv('lanternwick_and_Co_clean.csv', index=False)
```
